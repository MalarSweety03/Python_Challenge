Q1. Create a tuple named fruits containing "apple", "banana",
and "orange". Print the tuple.

```python
fruits = ("apple", "banana", "orange")
print(fruits)
```

Q2. Print the first element and the last element using indexing.

fruits = ("apple", "banana", "orange", "mango")

```python
fruits = ("apple", "banana", "orange", "mango")
print(fruits[0])
print(fruits[-1])
```

Q3. Print ("banana", "orange", "mango") using tuple slicing.

fruits = ("apple", "banana", "orange", "mango", "grapes")

```python
fruits = ("apple", "banana", "orange", "mango", "grapes")
print(fruits[1:4])
```

Q4. Count how many times the number 20 appears in the tuple.

numbers = (10, 20, 30, 20, 40, 20, 50)

```python
numbers = (10, 20, 30, 20, 40, 20, 50)
print(numbers.count(20))
```

Q5. Find and print the index position of the number 40.

numbers = (10, 20, 30, 40, 50)

```python
numbers = (10, 20, 30, 40, 50)
print(numbers.index(40))
```

Q6. Pack the three variables into a tuple named student and print it.

name = "Priya"
age = 22
course = "Data Analytics"

```python
name = "Priya"
age = 22
course = "Data Analytics"
student = (name, age, course)
print("Packed tuple:", student)
```

Q7. Unpack the tuple into three variables and print each variable.

student = ("Priya", 22, "Data Analytics")

```python
student = ("Priya", 22, "Data Analytics")
name, age, course = student
print("name:", name)
print("age:", age)
print("course", course)
```

Q8. Try changing the second element from 20 to 200 and print the tuple.

numbers = (10, 20, 30, 40, 50)

```python
numbers = (10, 20, 30, 40, 50)
elements = list(numbers)
elements[1] = 200
numbers = tuple(elements)
print(numbers)
```
