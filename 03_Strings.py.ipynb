Create a variable called name and assign the value "Python" to it.

Print the string and its length using len().

```python
name = "Python"
print(len(name))

```

Given text = "Python", print the first and last characters using indexing.

```python
text = "Python"
print(text[0])
print(text[5])
```

Given text = "Python", print the first three characters using slicing.

```python
text = "Python"
print(text[0:3])
```

Given text = "Python Programming", print only "Programming" using slicing.

```python
text = "Python Programming"
print(text[7:18])
```

Given text = "python", convert the string to uppercase using a string method.

```python
text = "python"
print(text.upper())
```

Given text = "PYTHON", convert the string to lowercase using a string method.

```python
text = "PYTHON"
print(text.lower())
```

Remove the leading and trailing spaces using a string method and print the result.

```python
text = "  Hello Python  "
print(text.strip())
```

Combine the two strings into "Data Analyst" using a string method.

```python
first = "Data"
second = "Analyst"
print(" ".join([first,second]))
```

Replace "Java" with "Python" using a string method and print the result.

```python
text = "I love Java"
print(text.replace("Java","Python"))
```

Count how many times the letter "a" appears in the string and print the result.

```python
text = "banana"
print(text.count("a"))
```

```python
text = "Python Programming"
print(text.index("Pro"))
```

Check whether the string starts with "Python" and print the result using a string method.

```python
text = "Python Programming"
print(text.startswith("Python"))
```

Check whether the string ends with ".csv" using a string method and print the result.

```python
text = "report.csv"
print(text.endswith(".csv"))
```

Reverse the string using slicing and print the result.

```python
text = "Python"
print(text[::-1])
```

Check whether the string contains only digits using a string method. Print the result.

```python
text = "12345"
print(text.isdigit())
```

Check whether the string contains only letters using a string method and print the result.

```python
text = "Python"
print(text.isalpha())
```

Check whether the string contains only letters and numbers using a string method. Print the result.

```python
text = "Data123"
print(text.isalnum())
```
