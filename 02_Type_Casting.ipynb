Basic Data Types

Print the data type of each variable using type().

```python
age = 25
price = 99.5
name = "Malar"
is_student = True
print(type(age))
print(type(price))
print(type(name))
print(type(is_student))
```

Type Casting

Convert num from a string to an integer, then print the converted value and its data type.

```python
num = "100"
num = int("100")
print(type(num))
```

Convert price into an integer and print the converted value.

```python
price = 49.99
price = int(49.99)
print(price)
```

Convert value into a string and print its data type using type().

```python
value = 25
value = str(25)
print(type(value))
```

Convert num into a Boolean and print the result.

```python
num = 0
num = bool(0)
print(num)
```
