Q12. Create two sets:

a = {10, 20}

b = {10, 20, 30, 40}
Check whether a is a subset of b using issubset().
Check whether b is a superset of a using issuperset().
Print both results.

```python
a = {10, 20}

b = {10, 20, 30, 40}
print("a issubset of b:", a.issubset(b))
print("b issuperset of a", b.issuperset(a))
```
